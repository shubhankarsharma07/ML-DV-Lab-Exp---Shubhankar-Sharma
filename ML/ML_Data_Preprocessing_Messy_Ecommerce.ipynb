```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Machine-learning1.0/Files/messy_ecommerce_data.csv")
```

```python
df.isnull().sum()
```

```python
df.fillna({
    "Transaction_ID": df["Transaction_ID"].mode()[0],
    "Transaction_Amount": df["Transaction_Amount"].mode()[0],
    "Transaction_Date": df["Transaction_Date"].mode()[0],
    "Customer_Name": df["Customer_Name"].mode()[0],
    "Email_Address": df["Email_Address"].mode()[0],
    "City": df["City"].mode()[0],
    "Is_Active": df["Is_Active"].mode()[0],
    "Product_Category": df["Product_Category"].mode()[0]
}, inplace=True)
```

```python
print(f"Customer Age : {df["Customer_Age"].mean()}")
print(f"Customer Rating :{df["Customer_Rating"].mean()}")
```

There are two method to find outliers one with IQR(Inter Quartile Range) and second is Z score method

# **IQR** **METHOD**

```python
data = np.array(df["Customer_Age"])
q1,q3 = np.percentile(data,[25,75])
iqr = q3-q1
lower_bound = q1 - (1.5*iqr)
upper_bound = q3 + (1.5*iqr)
print(f"Lower bound : {lower_bound}")
print(f"Upper bound : {upper_bound}")
outliers = data[(data<lower_bound) | (data>upper_bound)]
print(f"Outliers : {outliers}")
```

```python
df[df['Customer_Age']>100]
```

# **Z Score Method**

```python
from scipy import stats
zscores = np.abs(stats.zscore(data))
outliers = data[zscores>3]
print(f"Outliers : {outliers}")
```

```python
df[df['Customer_Rating'] > 10.0]
```

**So as Customer rating contains outlier i.e., greater than 10.0 whereas and customer age contains extremely ;arge value that can affect our central tendency if we calculate mean of thae data and useless to provide us the insights of data so for that we'll use median for this**

```python
df.fillna({
    "Customer_Age": df["Customer_Age"].median(),
    "Customer_Rating": df["Customer_Rating"].median()
},inplace=True)
```

```python
df.isnull().sum()
```

```python
pd.to_datetime(
    df["Transaction_Date"],
    errors = "coerce",
    dayfirst=True
)
```

```python
df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"], errors='coerce', dayfirst=True)
```

```python
df.dropna(subset=['Transaction_Date'])
```

```python
df['Transaction_Amount'] = df['Transaction_Amount'].str.replace("$","",regex=False)
df['Transaction_Amount'] = df['Transaction_Amount'].replace("Error",np.nan)
df['Transaction_Amount'] = df['Transaction_Amount'].astype(float)
```

```python
IA = {
    "Yes" : True,
    "No" : False,
    "True" : True,
    "False" : False,
    "1" : True,
    "0" : False
}
df["Is_Active"] = df["Is_Active"].map(IA)
```
